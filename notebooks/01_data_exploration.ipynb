# 01 — Data Exploration

Explore the `th1nhng0/vietnamese-legal-documents` dataset:
- Dataset configs: metadata, content, relationships
- Document statistics: counts, field distributions, date ranges
- Relationship graph structure
- Sample raw and cleaned text

```python
import sys
sys.path.insert(0, '..')

from dotenv import load_dotenv
load_dotenv('../.env')

from configs import config
```

## 1. Load a small sample

```python
from src.ingestion.loader import load_documents, load_relationships

# First run downloads the full dataset to ../data/raw/ (one-time, ~few minutes)
# Subsequent runs load from local parquet files instantly
docs = load_documents(config, sample_size=100)
print(f"Loaded {len(docs)} documents")
print("\nSample doc metadata:")
print(docs[0].metadata)
```

## 2. Metadata field analysis

```python
import pandas as pd

df = pd.DataFrame([d.metadata for d in docs])
print(df.dtypes)
print("\nDocument types:")
print(df['doc_type'].value_counts())
print("\nTop authorities:")
print(df['authority'].value_counts().head(10))
```

## 3. Raw vs Cleaned text comparison

```python
from src.ingestion.cleaner import clean_document

raw_doc = docs[0]
clean_doc = clean_document(raw_doc)

print("RAW (first 500 chars):")
print(raw_doc.page_content[:500])
print("\nCLEANED (first 500 chars):")
print(clean_doc.page_content[:500])
```

## 4. Chunking demonstration

```python
from src.ingestion.cleaner import clean_documents
from src.ingestion.chunker import chunk_documents

cleaned = clean_documents(docs[:10])
chunks = chunk_documents(cleaned, config)

print(f"10 docs → {len(chunks)} chunks")
print(f"Avg chunk length: {sum(len(c.page_content) for c in chunks) / len(chunks):.0f} chars")
print("\nSample chunk:")
print(chunks[0].page_content[:400])
print("\nChunk metadata:")
print(chunks[0].metadata)
```

## 5. Relationship graph sample

```python
import pandas as pd

rels = load_relationships(config)
print(f"Total relationships: {len(rels)}")

rels_df = pd.DataFrame(rels)
print("\nRelationship types:")
print(rels_df['relationship'].value_counts())
print("\nSample row:")
print(rels_df[['doc_id', 'other_doc_id', 'relationship']].head(5))
```
