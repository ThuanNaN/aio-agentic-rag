```python
import os
from dotenv import load_dotenv
load_dotenv()

CHROMA_HOST = "0.0.0.0"
CHROMA_PORT = 8000
TENANT = "default_tenant"
DATABASE = "default_database"
```

## Get Collections — chromadb client

```python
import chromadb

client = chromadb.HttpClient(host=CHROMA_HOST, port=CHROMA_PORT)
client.heartbeat()
```

```python
collections = client.list_collections()
for col in collections:
    print(f"name={col.name}  id={col.id}  count={client.get_collection(col.name).count()}")
```

## Get Collections — REST API v2

```python
import httpx

base_url = f"http://{CHROMA_HOST}:{CHROMA_PORT}"
url = f"{base_url}/api/v2/tenants/{TENANT}/databases/{DATABASE}/collections"
resp = httpx.get(url)
resp.raise_for_status()
resp.json()
```
